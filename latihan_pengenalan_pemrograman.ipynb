```python

# ===== PROFIL PRIBADI =====
profil_nama = "Annisa Nur Hapidah"
profil_pendidikan = "SMK Negeri 1 Cimahi"
profil_jurusan = "Rekayasa Perangkat Lunak"
profil_domisili = "Margaasih"
profil_profesi = "IT Support dan Admin"
profil_impian = "Data Analyst"

daftar_hobi = [
    "Membaca Buku",
    "Mendengarkan Musik",
    "Berlari"
]

# ===== MENYIMPAN DATA =====
profil = {
    "Nama Lengkap": profil_nama,
    "Asal Sekolah": profil_pendidikan,
    "Program Keahlian": profil_jurusan,
    "Tempat Tinggal": profil_domisili,
    "Pekerjaan": profil_profesi,
    "Hobi": daftar_hobi,
    "Cita-cita": profil_impian
}

# ===== MENAMPILKAN PROFIL =====
def tampilkan_profil():
    print("\n*** PROFIL SAYA ***")
    print("-" * 35)

    for informasi, isi in profil.items():
        if informasi != "Hobi":
            print(f"{informasi:<20}: {isi}")

    print("\nAktivitas yang Saya Sukai:")
    for nomor, aktivitas in enumerate(profil["Hobi"], 1):
        print(f"{nomor}) {aktivitas}")

    print("\nTarget Karier:")
    print(f"Saya ingin berkarier sebagai {profil['Cita-cita']}.")

    print("-" * 35)
    print("Senang bisa memperkenalkan diri!")

# ===== EKSEKUSI PROGRAM =====
tampilkan_profil()
```
